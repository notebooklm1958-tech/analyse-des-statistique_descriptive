## 11. Moyenne du chiffre d’affaires

```markdown
Nous calculons ici la valeur moyenne de la colonne du chiffre d'affaires (revenue).
```

````python
mean = df['revenue ($ millions)'].mean()
print(mean)
````

```markdown
**Interprétation du résultat :** Le nombre affiché représente le chiffre d'affaires moyen (en millions de dollars) d'une entreprise du Fortune 500 en 2019. Cette valeur est fortement influencée par les géants du sommet du classement.
```

## 12. Médiane du chiffre d’affaires

```markdown
Cette cellule calcule la valeur médiane, c'est-à-dire le point milieu où 50% des entreprises gagnent plus et 50% gagnent moins.
```

````python
median = df['revenue ($ millions)'].median()
print(median)
````

```markdown
**Interprétation du résultat :** La médiane sépare le classement en deux parties égales. La moitié des entreprises du Fortune 500 ont un chiffre d'affaires inférieur à cette valeur, et l'autre moitié a un chiffre d'affaires supérieur.
```

## 13. Mode du chiffre d’affaires

```markdown
Nous recherchons ici le mode, qui représente la valeur du chiffre d'affaires la plus fréquemment observée dans les données.
```

````python
mode = df['revenue ($ millions)'].mode()
print(mode)
````

```markdown
**Interprétation du résultat :** Le résultat montre la ou les valeurs de chiffre d'affaires qui apparaissent le plus souvent. Si toutes les valeurs sont uniques, le mode retourné sera simplement la liste complète ou la première valeur du classement.
```

## 14. Valeur minimale

```markdown
Cette cellule extrait le chiffre d'affaires le plus bas (valeur minimale) enregistré dans le classement.
```

````python
df['revenue ($ millions)'].min()
````

```markdown
**Interprétation du résultat :** C'est le chiffre d'affaires de la 500ème entreprise du classement, représentant le ticket d'entrée minimal pour faire partie du prestigieux Fortune 500 en 2019.
```

## 15. Valeur maximale

```markdown
Cette cellule extrait le chiffre d'affaires le plus élevé (valeur maximale) enregistré dans le classement.
```

````python
df['revenue ($ millions)'].max()
````

```markdown
**Interprétation du résultat :** C'est le chiffre d'affaires le plus colossal du classement (généralement détenu par le numéro 1, comme Walmart), exprimé en millions de dollars.
```

## 16. Étendue (Plage de variation)

```markdown
L'étendue (ou plage de variation) est calculée en soustrayant la valeur minimale de la valeur maximale.
```

````python
df['revenue ($ millions)'].max() - df['revenue ($ millions)'].min()
````

```markdown
**Interprétation du résultat :** Cette valeur montre l'écart gigantesque qui sépare la première entreprise (la plus riche) de la dernière du classement Fortune 500.
```

## 17. Variance

```markdown
La variance mesure la dispersion des valeurs du chiffre d'affaires autour de leur moyenne géométrique.
```

````python
df['revenue ($ millions)'].var()
````

```markdown
**Interprétation du résultat :** Une variance très élevée indique que les chiffres d'affaires des entreprises sont extrêmement dispersés et éloignés de la moyenne, traduisant une forte disparité de taille entre les membres du classement.
```

## 18. Écart-type

```markdown
L'écart-type (racine carrée de la variance) exprime la dispersion moyenne des données dans la même unité que la variable d'origine (en millions de dollars).
```

````python
df['revenue ($ millions)'].std()
````

```markdown
**Interprétation du résultat :** L'écart-type quantifie l'écart moyen par rapport au chiffre d'affaires moyen. Plus il est grand, plus les écarts de richesse entre les entreprises sont marqués.
```

## 19. Médiane (Q2, 50e percentile)

```markdown
Le deuxième quartile (Q2) correspond exactement à la médiane (50e percentile) calculée précédemment.
```

````python
Q2 = df['revenue ($ millions)'].quantile(0.5)
Q2
````

```markdown
**Interprétation du résultat :** Cette valeur est rigoureusement identique à la médiane calculée à l'étape 12. Elle confirme le point d'équilibre à 50% de l'échantillon.
```

## 20. Troisième quartile (Q3, 75e percentile)

```markdown
Le troisième quartile (Q3) indique le seuil en dessous duquel se situent 75% des entreprises du classement.
```

````python
Q3 = df['revenue ($ millions)'].quantile(0.75)
Q3
````

```markdown
**Interprétation du résultat :** Ce nombre indique que 75% des entreprises ont un chiffre d'affaires inférieur à ce seuil, tandis que seulement 25% (les plus grandes) se situent au-dessus.
```

## 21. Premier quartile (Q1, 25e percentile)

```markdown
Le premier quartile (Q1) indique le seuil en dessous duquel se situent 25% des entreprises.
```

````python
Q1 = df['revenue ($ millions)'].quantile(0.25)
Q1
````

```markdown
**Interprétation du résultat :** Ce nombre indique que seulement 25% des entreprises se situent sous ce seuil de chiffre d'affaires, les 75% restants gagnant plus.
```

## 22. Écart interquartile (IQR)

```markdown
L'écart interquartile (IQR) est la différence entre Q3 et Q1. Il mesure la dispersion de la moitié centrale des données.
```

````python
IQR = Q3 - Q1
IQR
````

```markdown
**Interprétation du résultat :** Cet écart représente la hauteur du bloc central des données (le cœur du classement). Il indique la dispersion des 50% d'entreprises situées au milieu du classement.
```

## 23. Diagramme en boîte (Boxplot)

```markdown
Nous générons un diagramme en boîte (boxplot) pour visualiser la distribution des données et repérer graphiquement les valeurs aberrantes (outliers).
```

````python
plt.boxplot(df['revenue ($ millions)'])
plt.show()
````

```markdown

Ces points représentent les entreprises exceptionnellement riches (les valeurs aberrantes ou outliers) qui dominent largement le reste du groupe.
```

## 24. Coefficient d'asymétrie (Skewness)

```markdown
Le coefficient d'asymétrie (skewness) indique si la distribution des données est symétrique ou étirée vers un côté (asymétrie positive ou négative).
```

````python
df['revenue ($ millions)'].skew()
````

```markdown
**Interprétation du résultat :** Un coefficient d'asymétrie largement supérieur à 0 (positif) confirme que la distribution est fortement étirée vers la droite. Cela signifie qu'il y a une grande majorité de "petites" entreprises du Fortune 500 et un très petit nombre de géants aux revenus démesurés.
```

## 25. Distribution du chiffre d’affaires

```markdown
Cette cellule trace un histogramme combiné à une estimation de la densité (KDE) pour visualiser la forme globale de la distribution des revenus.
```

````python
data = df['revenue ($ millions)']
sns.distplot(data, bins=10, hist=True, kde=True, label = 'revenue ($ millions)')
````

```markdown
**Interprétation du résultat :** La courbe de distribution montre une forme en "L" ou une décroissance très rapide. La grande majorité des observations sont concentrées près de l'axe vertical gauche, ce qui illustre visuellement l'inégalité des revenus.
```

## 26. Coefficient d'aplatissement (Kurtosis)

```markdown
Le coefficient d'aplatissement (kurtosis) mesure le degré de concentration des données autour de la moyenne (comparé à une distribution normale).
```

````python
df['revenue ($ millions)'].kurt()
````

```markdown
**Interprétation du résultat :** Un score d'aplatissement (kurtosis) très élevé (supérieur à 3) indique une distribution dite "leptokurtique". Elle possède un pic très prononcé et des queues épaisses, ce qui caractérise la présence de valeurs extrêmes très marquées (les leaders du classement).
```

**Note de compatibilité :** `sns.distplot` est conservé tel qu’il apparaît dans la vidéo. Cette fonction est obsolète dans les versions récentes de Seaborn ; si nécessaire, remplacez-la par `sns.histplot(data, bins=10, stat='density', kde=True, label='revenue ($ millions)')`.
